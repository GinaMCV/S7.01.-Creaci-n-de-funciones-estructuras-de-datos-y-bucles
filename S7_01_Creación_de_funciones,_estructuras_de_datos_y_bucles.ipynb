## **Ejercicio 1: Selección de Personal**
Se ha abierto una nueva oferta laboral interna y la empresa ha decidido que sólo las primeras 10 personas que se inscriban podrán participar en el proceso de selección.

Desarrollar una herramienta sencilla en Python que actúe como formulario de inscripción y sistema de clasificación inicial.Este formulario debe permitir introducir, para cada candidato/a:

1. Nombre y Apellidos
2. Teléfono de contacto
3. Años de experiencia dentro de la empresa
4. Conocimiento de Catalán (0-3 puntos)
- 0 = sin conocimiento
- 1 = hasta nivel B1
- 2 = nivel B2
- 3 = C1 o superior
5. Dominio de Excel (0-3 puntos)
- 0 = nivel básico
- 1 = uso de funciones
- 2 = tablas dinámicas
- 3 = Visual Basic y Macros
6. Título académico (0-3 puntos)
- 0 = ESO
- 1 = FP
- 2 = Universidad
- 3 = Máster o superior


`obtener_datos()`
- No recibe parámetros de entrada.
- Solicita por teclado los siguientes datos: nombre y apellidos, teléfono de contacto, años de experiencia dentro de la empresa, conocimiento de catalán, dominio de Excel y titulación académica.
- Guarda todos los datos introducidos por el usuario.
- Devuelve un diccionario que contiene los seis valores, utilizando como claves: `"nombre"`, `"telefono"`, `"experiencia"`, `"catalan"`, `"excel"` y `"titulacion"`.
- Los valores introducidos mediante `input()` se almacenan inicialmente como cadenas de texto `(str)`.

```python
def obtener_datos():
  nombre = input("Nombre y Apellidos:  ")
  telefono = input("Teléfono de contacto:  ")
  experiencia = input("Años de experiencia dentro de la empresa:  ")
  catalan = input("Conocimiento de Catalán (0-3 puntos):  ")
  excel = input("Dominio de Excel (0-3 puntos): ")
  titulacion = input("Título académico (0-3 puntos): ")
  return {
        "nombre": nombre,
        "telefono": telefono,
        "experiencia": experiencia,
        "catalan": catalan,
        "excel": excel,
        "titulacion": titulacion
    }
```


`validar_datos(datos)`

- Recibe como parámetro un diccionario `datos` que contiene la información de un candidato.
- Intenta convertir a números enteros (`int`) los valores correspondientes a experiencia, catalán, Excel y titulación.
- Utiliza un bloque `try-except` para controlar posibles errores de conversión o de tipo.
- Comprueba que:
  - El nombre no esté vacío y que contenga únicamente letras y espacios.
  - El teléfono tenga exactamente 9 caracteres y que todos sean dígitos.
  - Los años de experiencia sean mayores o iguales a 0.
  - Las puntuaciones de catalán, Excel y titulación estén comprendidas entre 0 y 3.
- Si todos los datos son válidos, devuelve el diccionario `datos` con los valores numéricos convertidos a enteros.
- Si algún dato no es válido o se produce un error durante la conversión o validación, devuelve `False`.


```python
def validar_datos(datos):
    try:
        datos["experiencia"] = int(datos["experiencia"])
        datos["catalan"] = int(datos["catalan"])
        datos["excel"] = int(datos["excel"])
        datos["titulacion"] = int(datos["titulacion"])
        if (
            datos["nombre"].strip() != "" and
            datos["nombre"].replace(" ", "").isalpha() and
            len(datos["telefono"]) == 9 and
            datos["telefono"].isdigit() and
            datos["experiencia"] >= 0 and
            datos["catalan"] in range(4) and
            datos["excel"] in range(4) and
            datos["titulacion"] in range(4)
        ):
            return datos

    except (ValueError, TypeError):
        return False

    return False
```

Asegurándose de que toda la información introducida es correcta y entra dentro de los valores permitidos. A partir de estos datos, el programa deberá calcular una puntuación total aplicando el peso de la experiencia (años * 1.5) y sumando los puntos de catalán, excel y titulación



`calcular_puntuacion(datos)`

- Recibe como parámetro un diccionario `datos` que contiene la información del candidato.
- Obtiene del diccionario los valores de `"experiencia"`, `"catalan"`, `"excel"` y `"titulacion"`.
- Calcula la puntuación total del candidato aplicando la siguiente fórmula:
  - **puntuación = experiencia × 1.5 + catalán + Excel + titulación**
- Devuelve la puntuación total calculada.


```python
def calcular_puntuacion(datos):
    puntuacion = (
        datos["experiencia"] * 1.5
        + datos["catalan"]
        + datos["excel"]
        + datos["titulacion"]
    )
    return puntuacion
```

y a continuación clasificar automáticamente a cada persona según los umbrales establecidos por la empresa:

- Prioridad Alta → puntuación ≥ 12
- Candidato adecuado → 8 ≤ puntuación < 12
- No prioritario → puntuación < 8


`clasificar_candidato(puntuacion)`
- Recibe como parámetro la puntuación total obtenida por el candidato.
- Si la puntuación es mayor o igual a 12, clasifica al candidato como
  - **"Prioridad alta"**.
* Si la puntuación es mayor o igual a 8 pero menor que 12, clasifica al candidato como
  - **"Candidato adecuado"**.
* Si la puntuación es menor que 8, clasifica al candidato como
  - **"No prioritario"**.
* Devuelve una cadena de texto con la clasificación correspondiente al candidato.


```python
def clasificar_candidato(puntuacion):
    if puntuacion >= 12:
        return "Prioridad alta"
    elif puntuacion >= 8:
        return "Candidato adecuado"
    else:
        return "No prioritario"
```

Cada vez que alguien completo en el formulario, debe mostrar su puntaje individual.


`mostrar_individual(nombre, puntuacion, clasificacion)`

- Recibe como parámetros el nombre del candidato, su puntuación y su clasificación.
- No devuelve ningún valor, únicamente muestra por pantalla el nombre, la puntuación obtenida y la clasificación asignada del candidato.


```python
def mostrar_individual(nombre, puntuacion, clasificacion):
  print(
        f"\nCandidato: {nombre}  | "
        f"Puntuación: {puntuacion}  | "
        f"Clasificación: {clasificacion} | ")
```




 Cuando ya se hayan inscrito exactamente diez personas, la herramienta debe dejar de aceptar nuevos registros y generar un informe final en pantalla, ordenado primero según la prioridad y, en caso de empate, por la puntuación obtenida. Este informe debe incluir el nombre, el teléfono y la puntuación de cada candidato para que el departamento de RRHH pueda contactarles rápidamente.



`mostrar_informe(candidatos)`

- Recibe como parámetro una lista de diccionarios con la información de los candidatos.
- Define un orden de prioridad para las diferentes clasificaciones:
  - **"Prioridad alta"** → prioridad 0.
  - **"Candidato adecuado"** → prioridad 1.
  - **"No prioritario"** → prioridad 2.
-  Ordena la lista de candidatos según su clasificación, colocando primero los candidatos con mayor prioridad.
- En caso de que dos candidatos tengan la misma clasificación, los ordena por puntuación de mayor a menor.
- Muestra por pantalla un encabezado con el título **"INFORME FINAL"**.
- Recorre la lista ordenada y muestra para cada candidato su nombre, teléfono y puntuación.
- No devuelve ningún valor, únicamente muestra el informe por pantalla.


```python
def mostrar_informe(candidatos):
    prioridad = {
        "Prioridad alta": 0,
        "Candidato adecuado": 1,
        "No prioritario": 2
    }
    candidatos.sort(
        key=lambda candidato: (prioridad[candidato["clasificacion"]],
                        -candidato["puntuacion"])
    )
    print("\n" + "=" * 72)
    print("=" * 28, "INFORME FINAL", "=" * 29)
    print("=" * 72)
    for candidato in candidatos:
        print(
            f'{candidato["nombre"]} - '
            f'{candidato["telefono"]} - '
            f'{candidato["puntuacion"]}'
        )
```

#Funcion principal



`main_seleccion()`

-  Crea una lista vacía llamada `candidatos` para almacenar los candidatos válidos.
-  Muestra por pantalla un encabezado con el título **"CANDIDATOS"**.
-  Solicita los datos de cada candidato mediante la función `obtener_datos()`, que devuelve un diccionario.
-  Continúa solicitando candidatos mientras la lista contenga menos de 10 candidatos válidos.
-  Valida los datos mediante la función `validar_datos()`.
-  Si los datos no son válidos, muestra un mensaje indicando que los datos del candidato no son válidos y vuelve a solicitar los datos.
-  Si los datos son válidos, calcula la puntuación del candidato mediante `calcular_puntuacion()`.
-  Clasifica al candidato según su puntuación mediante `clasificar_candidato()`.
-  Crea un nuevo diccionario con el nombre, teléfono, puntuación y clasificación del candidato.
-  Añade el diccionario del candidato válido a la lista `candidatos`.
-  Muestra individualmente los datos del candidato mediante `mostrar_individual()`.
-  Cuando se han registrado 10 candidatos válidos, muestra el informe final mediante `mostrar_informe()`.
- No devuelve ningún valor; su función es gestionar el proceso completo de selección y mostrar los resultados.



```python
def main_seleccion():
    candidatos = []
    print("\n" + "=" * 72)
    print("=" * 30, "CANDIDATOS", "=" * 30)
    print("=" * 72)
    while len(candidatos) < 10:
        datos = obtener_datos()
        if not validar_datos(datos):
            print(f'\nDatos no válidos para {datos["nombre"]}')
            continue
        puntuacion = calcular_puntuacion(datos)
        clasificacion = clasificar_candidato(puntuacion)
        candidato = {
            "nombre": datos["nombre"],
            "telefono": datos["telefono"],
            "puntuacion": puntuacion,
            "clasificacion": clasificacion
        }
        candidatos.append(candidato)
        mostrar_individual(
            datos["nombre"],
            puntuacion,
            clasificacion
        )
    mostrar_informe(candidatos)
```

Dispones de un conjunto de datos con información básica de 10 clientes:

- id_client
- nombre

```python
clients = [
    {"id_client": 1, "nom": "Laura Serra", "edat": 23, "grup_edat": "18-25", "codi_comportament": 4},
    {"id_client": 2, "nom": "Marc Vidal", "edat": 35, "grup_edat": "26-40", "codi_comportament": 3},
    {"id_client": 3, "nom": "Ana López", "edat": 29, "grup_edat": "26-40", "codi_comportament": 2},
    {"id_client": 4, "nom": "Joan Riera", "edat": 19, "grup_edat": "18-25", "codi_comportament": 2},
    {"id_client": 5, "nom": "Mónica Pérez", "edat": 47, "grup_edat": "41-60", "codi_comportament": 4},
    {"id_client": 6, "nom": "Luis García", "edat": 52, "grup_edat": "41-60", "codi_comportament": 1},
    {"id_client": 7, "nom": "Pilar Sánchez", "edat": 61, "grup_edat": "60+", "codi_comportament": 3},
    {"id_client": 8, "nom": "Eva Martín", "edat": 38, "grup_edat": "26-40", "codi_comportament": 0},
    {"id_client": 9, "nom": "Diego Romero", "edat": 24, "grup_edat": "18-25", "codi_comportament": 1},
    {"id_client": 10, "nom": "Núria Costa", "edat": 33, "grup_edat": "26-40", "codi_comportament": 4}
]
```

El departamento de CRM ha definido manualmente:

1. Códigos de comportamiento
- 0 → Navega pero no añade productos
- 1 → Agregar productos pero no llega al checkout
- 2 → Llega al checkout pero no paga
- 3 → Cesta de compra abandonado hace menos de 24 h
- 4 → Cesta de compra abandonado hace más de 24 h
2. Factores según código de comportamiento
- Código 0 → 0.10
- Código 1 → 0.25
- Código 2 → 0.50
- Código 3 → 0.75
- Código 4 → 0.90

`obtenir_segment_i_factor(codigo_comportamiento)`
- La función recibe como parámetro codigo_comportamiento y utiliza este valor como clave para acceder al diccionario.
- Se crea un diccionario comportamientos que contiene la información asociada a cada código de comportamiento.
- La función permite obtener de forma estructurada el segmento, el mensaje y el factor de código que posteriormente se utilizarán para calcular el riesgo del cliente.
- Se devuelve un diccionario con la información correspondiente al código de comportamiento recibido.


```python
def obtenir_segment_i_factor(codigo_comportamiento):
    comportamientos = {
        0: {"segmento": "Navega pero no añade productos", "mensaje": "Descubre nuestros productos.", "factor_codigo": 0.10},
        1: {"segmento": "Añade productos pero no llega al checkout", "mensaje": "Tus productos te están esperando.", "factor_codigo": 0.25},
        2: {"segmento": "Llega al checkout pero no paga", "mensaje": "Tu compra está a un paso. ¡Finalízala!", "factor_codigo": 0.50 },
        3: {"segmento": "Cesta abandonada hace menos de 24 h", "mensaje": "Recupera tu cesta y completa tu compra.", "factor_codigo": 0.75},
        4: {"segmento": "Cesta abandonada hace más de 24 h", "mensaje": "¡No pierdas tus productos! Recupera tu cesta.", "factor_codigo": 0.90}
    }
    return comportamientos[codigo_comportamiento]
```


3.  Factores según grupo de edad
- 18–25 → 1.20
- 26–40 → 1.00
- 41–60 → 0.85
- 60+ → 0.70

`obtenir_factor_edat(grupo_edad)`
- La función recibe como parámetro grupo_edad y utiliza este valor como clave para acceder al diccionario.
- Se crea un diccionario factores_edad que contiene los diferentes grupos de edad y el factor correspondiente a cada uno.
- El factor obtenido se utilizará posteriormente junto con el factor de comportamiento para calcular el riesgo final del cliente.
- Se devuelve el factor de edad correspondiente al grupo de edad recibido.

```python
def obtenir_factor_edat(grupo_edad):
    factores_edad = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70
    }
    return factores_edad[grupo_edad]
```

4.  Fórmula del riesgo final
- risc_final = factor codi * factor edat

`calcular_risc(factor_codigo, factor_edad)`
- La función recibe como parámetros el factor de comportamiento `(factor_codigo)` y el factor de edad `(factor_edad)`.
- Se multiplican ambos factores para obtener el riesgo final del cliente, el resultado de la multiplicación se almacena en la variable riesgo.
- devuelve el valor del riesgo calculado el cual permite determinar el nivel de riesgo de cada cliente teniendo en cuenta tanto su comportamiento de compra como su grupo de edad.

```python
def calcular_risc(factor_codigo, factor_edad):
  riesgo = factor_codigo * factor_edad
  return riesgo
```

- Clientes procesados

`processar_clients(clients)`
- Recibe como parámetro una lista de clientes.
- Se crea una lista vacía `clientes_procesados` donde se almacenará la información procesada de cada cliente.
- Se recorre la lista de clientes mediante un bucle for.
- Para cada cliente se obtiene su segmento, mensaje y factor de comportamiento mediante la función `obtenir_segment_i_factor()`.
- Se obtiene el factor de edad mediante la función `obtenir_factor_edat()`.
- Se calcula el riesgo final del cliente utilizando la `función calcular_risc()`, multiplicando el factor de comportamiento por el factor de edad.
- Se crea un nuevo diccionario `cliente_procesado` con los datos principales del cliente y la información calculada: `segmento`, `mensaje`, `factores` y `riesgo final`.
- Los valores de los factores y del riesgo final se redondean a tres decimales.
- Cada cliente procesado se añade a la lista `clientes_procesados`.
- Dvuelve la lista completa de clientes procesados.

```python
def processar_clients(clients):
    clientes_procesados = []
    for cliente in clients:
        comportamiento = obtenir_segment_i_factor(cliente["codi_comportament"])
        segmento = comportamiento["segmento"]
        mensaje = comportamiento["mensaje"]
        factor_codigo = comportamiento["factor_codigo"]
        factor_edad = obtenir_factor_edat(cliente["grup_edat"])
        riesgo_final = calcular_risc(factor_codigo,factor_edad)
        cliente_procesado = {
            "id_cliente": cliente["id_client"],
            "nombre": cliente["nom"],
            "edad": cliente["edat"],
            "grupo_edad": cliente["grup_edat"],
            "codigo_comportamiento": cliente["codi_comportament"],
            "segmento": segmento,
            "mensaje": mensaje,
            "factor_codigo": round(float(factor_codigo), 3),
            "factor_edad": round(float(factor_edad), 3),
            "riesgo_final": round(float(riesgo_final), 3)
        }
        clientes_procesados.append(cliente_procesado)

    return clientes_procesados
```

`mostrar_diccionari(clientes)`
- Recibe como parámetro una lista de clientes procesados.
- Se ordena la lista de clientes mediante la función `sorted()` según el valor de riesgo_final.
- Se utiliza `reverse=True` para ordenar los clientes de mayor a menor riesgo.
- Se muestra en pantalla un título indicando que se trata de los clientes procesados.
- Se crea una tabla con los encabezados correspondientes a la información de cada cliente: `ID`, `nombre`, `edad`, `grupo de edad`, `código de comportamiento`, `segmento`, `factor de código`, `factor de edad` y `riesgo final`.
- Mediante un bucle for se recorre los datos para cada cliente y se muestran sus datos en las diferentes columnas de la tabla utilizando `f-strings` y espacios de alineación para facilitar la lectura.

- No devuelve ningun valor, muestra los clientes procesados ordenados de mayor a menor según su riesgo final.

```python
def mostrar_diccionari(clientes):
    clientes_ordenados = sorted(
        clientes,
        key=lambda cliente: cliente["riesgo_final"],
        reverse=True
    )
    print("\n" + "=" * 150)
    print("=" * 64 + " CLIENTES PROCESADOS " + "=" * 65)
    print("=" * 150)
    print(
        f"{'ID':<4}"
        f"{'Nombre':<18}"
        f"{'Edad':<6}"
        f"{'Grupo':<9}"
        f"{'Código':<8}"
        f"{'Segmento':<48}"
        f"{'F. código':<11}"
        f"{'F. edad':<10}"
        f"{'Riesgo':<8}"
    )
    print("-" * 150)
    for cliente in clientes_ordenados:
        print(
            f"{cliente['id_cliente']:<4}"
            f"{cliente['nombre']:<18}"
            f"{cliente['edad']:<6}"
            f"{cliente['grupo_edad']:<9}"
            f"{cliente['codigo_comportamiento']:<8}"
            f"{cliente['segmento']:<48}"
            f"{cliente['factor_codigo']:<11}"
            f"{cliente['factor_edad']:<10}"
            f"{cliente['riesgo_final']:<8}"
        )
    print("=" * 150)
```

## **Ejercicio 1: Gestión de riesgo y priorización de pacientes**

Varios pacientes han enviado sus síntomas a través de un formulario online, tienes una lista con información real procedente del sistema:



```python
pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]
```

El objetivo es transformar esos datos en información útil para los médicos. A partir de los registros en bruto, debes:

1. Convertir cada string en una estructura limpia (diccionario o similar).


```python
def input_datos_pacientes(pacients, indice):
  """
    Obtiene los datos de un paciente utilizando su índice.

    Argumentos:
            (pacients (list)): Lista que contiene los datos de los pacientes.
            (indice (int)): Índice del paciente que se quiere obtener.

    Devuelve (str): Datos del paciente correspondiente al índice indicado.
  """
  return pacients[indice]


def obtener_datos_pacientes(pacients):
  """
    Obtiene y organiza los datos de todos los pacientes.

    Argumento (pacients (list)): Lista que contiene los datos de los pacientes.

    Devuelve (list): Lista de tuplas con el nombre, edad y síntomas de cada paciente.
  """
  lista_pacientes = []
  for indice in range(len(pacients)):
      datos = input_datos_pacientes(pacients, indice).split("|")
      nombre = datos[0]
      edad = datos[1]
      sintomas = datos[2]
      informacion = nombre, edad, sintomas
      lista_pacientes.append(informacion)
  return lista_pacientes
```

pero los datos llegan en un formato desordenado y poco estructurado , y el equipo médico necesita que los organices para priorizar el orden de atención según la gravedad.


2. Separar y normalizar la lista de síntomas:
- minúsculas
- eliminar espacios
- dividir correctamente por comas


```python
def normalizar_sintomas(sintomas):
  """
  Normaliza una cadena de síntomas para facilitar su procesamiento.

  Argumento (sintomas (str)): Cadena de texto con los síntomas separados por comas.

  Devuelve (list): Lista de síntomas normalizados en minúsculas,
     sin espacios innecesarios y con los espacios internos sustituidos por "_".

  """
  sintomas_normalizados = sintomas.lower().split(",")
  sintomas_normalizados = [s.strip() for s in sintomas_normalizados]
  sintomas_normalizados = [s.replace(" ", "_") for s in sintomas_normalizados]
  return sintomas_normalizados
```

 El departamento médico también te ha proporcionado el siguiente diccionario de niveles de riesgo por síntoma:

 `nivells_risc = {diccionario de niveles de riesgo por síntoma}`

```python
nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

3. Asignar un riesgo a cada síntoma utilizando el diccionario nivells_risc.

`calcular_riesgo_total(sintomas, nivells_risc)`

- Recibe como parámetros la lista de síntomas y el diccionario con los niveles de riesgo.
- Inicializa riesgo_total a 0.
- Recorre cada síntoma mediante un for.
- Busca el nivel de riesgo correspondiente a cada síntoma en nivells_risc.
- Suma cada nivel de riesgo al riesgo_total.
- Devuelve el riesgo total del paciente.

```python
def calcular_riesgo_total(sintomas, nivells_risc):
  """
    Calcula el riesgo total de un paciente a partir de sus síntomas.

    Argumentos:
        sintomas (list): Lista de síntomas normalizados del paciente.
        nivells_risc (dict): Diccionario que contiene el nivel de riesgo de cada síntoma.

    Devuelve:
        int: Riesgo total calculado a partir de los síntomas.
  """
  riesgo_total = 0
  for sintoma in sintomas:
      riesgo_total += nivells_risc[sintoma]
  return riesgo_total
```

4. Calcular el riesgo total de cada paciente sumando los valores.

```python
def niveles_urgencia(riesgo):
    """
    Determina el nivel de urgencia y el tiempo de atención
    según el riesgo total del paciente.

    Argumeto (riesgo (int)): Riesgo total calculado para el paciente.

    Devuelve (tuple): Nivel de urgencia y tiempo de atención.
    """
    niveles = {
        (1, 2): { "NIVEL": "Leve", "TIEMPO_ATENCION": "Hasta 120-240 minutos" },
        (3, 4): { "NIVEL": "Moderada", "TIEMPO_ATENCION": "Hasta 60 minutos" },
        (5, 6): { "NIVEL": "Urgente", "TIEMPO_ATENCION": "Hasta 30 minutos" },
        (7, 8): { "NIVEL": "Muy urgente", "TIEMPO_ATENCION": "Hasta 10-15 minutos" },
        (9, float("inf")): { "NIVEL": "Emergencia", "TIEMPO_ATENCION": "Atención inmediata" }
    }
    rango = next(
        rango
        for rango in niveles
        if rango[0] <= riesgo <= rango[1]
    )
    return (
        niveles[rango]["NIVEL"],
        niveles[rango]["TIEMPO_ATENCION"]
    )
```

Pacientes procesados

```python
def procesar_pacientes(pacients, nivells_risc):

    """
    Procesa los datos de los pacientes y calcula su nivel de riesgo.

    Argumentos:
        pacients (list): Lista con los datos de los pacientes.
        nivells_risc (dict): Diccionario con el nivel de riesgo de cada síntoma.

    Devuelve (list): Lista de pacientes procesados con sus síntomas normalizados,
        riesgo total, nivel de urgencia y tiempo de atención.
    """
    pacientes_procesados = []
    pacientes = obtener_datos_pacientes(pacients)
    for nombre, edad, sintomas in pacientes:
        sintomas_normalizados = normalizar_sintomas(sintomas)
        riesgo_total = calcular_riesgo_total(
            sintomas_normalizados,
            nivells_risc
        )
        nivel_urgencia, tiempo = niveles_urgencia(riesgo_total)
        paciente = {
            "nombre": nombre,
            "edad": int(edad),
            "sintomas": sintomas_normalizados,
            "riesgo_total": riesgo_total,
            "nivel_urgencia": nivel_urgencia,
            "tiempo_atencion": tiempo
        }
        pacientes_procesados.append(paciente)
    return pacientes_procesados
```


2. Calcular el precio final de la póliza con la fórmula dada.

El precio final debe calcularse con la fórmula:
- preu_final = preu_base * (1 + edat/100)

```python
def precio_final_cliente(precio_base, edad):
    """
    Calcula el precio final de la póliza según el precio base y la edad.

    Argumentos:
        precio_base (float): Precio base de la póliza.
        edad (int): Edad del cliente.

    Devuelve (float): Precio final de la póliza.
    """
    return precio_base * (1 + edad / 100)
```

```python
def procesar_clientes(clients, nivells_seguro):
    """
    Procesa los datos de los clientes y calcula la información de su póliza.

    Argumentos:
        clients (list): Lista con los datos de los clientes.
        nivells_seguro (dict): Diccionario con las categorías,
        precios base y probabilidades de fraude.

    Devuelve (list): Lista de clientes procesados con su categoría,
    precio base, precio final y probabilidad de fraude.
    """
    lista_clientes = []
    for cliente in clients:
        seguro = niveles_seguro(cliente["codi"], nivells_seguro)
        categoria = seguro["categoria"]
        precio_base = seguro["preu_base"]
        probabilidad_fraude = seguro["fraude"]
        edad = cliente["edat"]
        precio_final = precio_final_cliente(precio_base, edad)

        cliente_procesado = {
            "nombre": cliente["nom"],
            "edad": int(edad),
            "categoria": categoria,
            "precio_base": float(precio_base),
            "precio_final": round(float(precio_final), 3),
            "Probabilidad_fraude": float(probabilidad_fraude),
            "codigo": cliente["codi"]
        }
        lista_clientes.append(cliente_procesado)
    return lista_clientes
```

## **Ejercicio 1: Generador de contraseñas seguras**

Se ha detectado que la mayoría de trabajadores siguen utilizando contraseñas muy débiles como “Asdf1234”, fechas de cumpleaños o combinaciones fáciles de adivinar. Por este motivo, el responsable del departamento te ha pedido desarrollar un generador de contraseñas seguras , totalmente parametrizable y modular.

Requisitos del generador

Tienes que crear una función principal y todas las subfunciones necesarias para:

2. Decidir si deben incluir:
- Mayúsculas
- Minúsculas
- Números
- Signos especiales

```python
def obtener_caracteres(majuscules, minuscules, numeros, signes):

  """
    Obtiene una cadena de caracteres según las opciones seleccionadas.

    Args:
        majuscules (bool): Indica si se incluyen letras mayúsculas.
        minuscules (bool): Indica si se incluyen letras minúsculas.
        numeros (bool): Indica si se incluyen números.
        signes (bool): Indica si se incluyen signos especiales.

    Returns:
        str: Cadena con todos los caracteres seleccionados.
    """
  caracteres = ""
  if majuscules:
    caracteres += string.ascii_uppercase
  if minuscules:
    caracteres += string.ascii_lowercase
  if numeros:
    caracteres += string.digits
  if signes:
    caracteres += "!@#$%^&*()-_=+"
  return caracteres
```

3. Asegurarte de que al menos aparece un carácter de cada tipo seleccionado .

```python
def obtener_obligatorios(majuscules, minuscules, numeros, signes):
  """
    Genera una lista con un carácter obligatorio de cada tipo seleccionado.

    Args:
        majuscules (bool): Indica si se debe incluir una letra mayúscula.
        minuscules (bool): Indica si se debe incluir una letra minúscula.
        numeros (bool): Indica si se debe incluir un número.
        signes (bool): Indica si se debe incluir un signo especial.

    Returns:
        list: Lista con un carácter aleatorio de cada tipo seleccionado.
    """
  obligatorios = []
  if majuscules:
      obligatorios.append(random.choice(string.ascii_uppercase))
  if minuscules:
      obligatorios.append(random.choice(string.ascii_lowercase))
  if numeros:
      obligatorios.append(random.choice(string.digits))
  if signes:
      obligatorios.append(random.choice("!@#$%^&*()-_=+"))
  return obligatorios
```



1. Generar contraseñas de longitud variable.
4. Aleatorizar completamente la contraseña.
5. Devolver la contraseña generada.



```python

def crear_contrasenya(longitud, majuscules, minuscules, numeros, signes):
    """
    Genera una contraseña aleatoria con una longitud determinada.

    La contraseña debe contener como mínimo:
    - Una letra mayúscula.
    - Una letra minúscula.
    - Un número.
    - Un signo especial.

    Args:
        longitud (int): Longitud deseada para la contraseña.
        majuscules (bool): Incluye letras mayúsculas.
        minuscules (bool): Incluye letras minúsculas.
        numeros (bool): Incluye números.
        signes (bool): Incluye signos especiales.

    Returns:
        str: Mensaje con la contraseña generada.
        None: Si la longitud es menor de 4 caracteres.
    """

    if longitud < 4:
        print("\n" + "=" * 72)
        print("=" * 22, "ERROR", "=" * 43)
        print("=" * 72)
        print("La contraseña debe tener como mínimo 4 caracteres.")
        print("Debe contener una mayúscula, una minúscula, un número y un signo.")
        print("=" * 72)
        return None

    caracteres = obtener_caracteres(majuscules, minuscules, numeros, signes)
    obligatorios = obtener_obligatorios(majuscules, minuscules, numeros, signes)

    cantidad_restante = longitud - len(obligatorios)

    restantes = []

    for _ in range(cantidad_restante):
        restantes.append(random.choice(caracteres))

    contrasenya = obligatorios + restantes

    random.shuffle(contrasenya)

    return "La contraseña es: " + "".join(contrasenya)
```

La función principal del generador debe tener valores por defecto, exactamente los siguientes:

- majuscules=True
- minuscules=True
- numeros=True
- signes=False


## **Ejercicio 2: Procesamiento automático de datos deportivos**

Ahora formas parte de un equipo de analítica en una organización deportiva que gestiona datos de la liga catalana. Tu compañera de trabajo desea automatizar la lectura de su archivo histórico de resultados y extraer las estadísticas más importantes. Te ha pedido hacer un pequeño sistema en Python que procese el archivo 📄 "historic partits.txt"
\

```python
def leer_partits(partidos):
  """
    Lee los partidos desde un archivo de texto y guarda la información
    de cada partido en una lista de diccionarios.

    Args:
        partidos (str): Ruta o nombre del archivo que contiene los partidos.

    Returns:
        list: Lista de diccionarios con el equipo local, equipo visitante,
        goles del equipo local y goles del equipo visitante.
    """
  with open(partidos, "r", encoding="utf-8") as f:
      partits = []
      for linea in f:
          linea = linea.strip()
          if linea:
              equip_local, resultat, equip_visitant = linea.split("\t")
              gols_local, gols_visitant = resultat.split("-")
              partits.append({
                  "equip_local": equip_local,
                  "equip_visitant": equip_visitant,
                  "gols_local": int(gols_local),
                  "gols_visitant": int(gols_visitant)
              })
  return partits
```

Estadistica de goles


```python
def calcular_estadisticas(partidos):
  """
    Calcula las estadísticas de cada equipo a partir de los partidos.

    Para cada equipo calcula los goles a favor, goles en contra,
    partidos ganados, empatados, perdidos y los puntos obtenidos.

    Args:
        partidos (list): Lista de diccionarios con la información
        de los partidos.

    Returns:
        dict: Diccionario con las estadísticas de cada equipo.
    """
  equipos = {}
  for partido in partidos:
      local = partido["equip_local"]
      visitante = partido["equip_visitant"]
      goles_local = partido["gols_local"]
      goles_visitante = partido["gols_visitant"]

      if local not in equipos:
          equipos[local] = {
              "goles_favor": 0,
              "goles_contra": 0,
              "ganados": 0,
              "empatados": 0,
              "perdidos": 0,
              "puntos": 0
          }

      if visitante not in equipos:
          equipos[visitante] = {
              "goles_favor": 0,
              "goles_contra": 0,
              "ganados": 0,
              "empatados": 0,
              "perdidos": 0,
              "puntos": 0
          }

      equipos[local]["goles_favor"] += goles_local
      equipos[local]["goles_contra"] += goles_visitante

      equipos[visitante]["goles_favor"] += goles_visitante
      equipos[visitante]["goles_contra"] += goles_local

      if goles_local > goles_visitante:
        equipos[local]["ganados"] += 1
        equipos[local]["puntos"] += 3
        equipos[visitante]["perdidos"] += 1

      elif goles_local < goles_visitante:
        equipos[visitante]["ganados"] += 1
        equipos[visitante]["puntos"] += 3
        equipos[local]["perdidos"] += 1

      else:
        equipos[local]["empatados"] += 1
        equipos[visitante]["empatados"] += 1
        equipos[local]["puntos"] += 1
        equipos[visitante]["puntos"] += 1

  return equipos
```

Objetivos

Mediante un código modular, claro y documentado , tu programa debe calcular:

1. Qué equipo es el más goleador.

```python
def equipo_mas_goleador(equipos):
  """
    Obtiene el equipo que ha marcado más goles.

    Args:
        equipos (dict): Diccionario con las estadísticas de los equipos.

    Returns:
        tuple: Tupla con el nombre del equipo y sus estadísticas.
    """

  return max(
      equipos.items(),
      key=lambda equipo: equipo[1]["goles_favor"]
  )
```

2. Cuál es el más goleado.

```python
def equipo_mas_goleado(equipos):
  """
    Obtiene el equipo que ha recibido más goles.

    Args:
        equipos (dict): Diccionario con las estadísticas de los equipos.

    Returns:
        tuple: Tupla con el nombre del equipo y sus estadísticas.
    """
  return max(
      equipos.items(),
      key=lambda equipo: equipo[1]["goles_contra"]
  )
```

4. La clasificación final ordenada por puntos donde cada victoria suma 3 pts, cada empate 1 pts y cada derrota 0 pts.

```python
def ordenar_clasificacion(equipos):
  """
    Ordena los equipos de mayor a menor según los puntos obtenidos.

    Args:
        equipos (dict): Diccionario con las estadísticas de los equipos.

    Returns:
        list: Lista de tuplas con los equipos ordenados por puntos.
    """
  return sorted(
      equipos.items(),
      key=lambda equipo: equipo[1]["puntos"],
      reverse=True
  )
```


3. El número total de goles que ha marcado cada equipo.


```python
def mostrar_resultados(equipos):
  """
    Muestra por pantalla las estadísticas principales de la liga
    y la clasificación final de los equipos.

    Muestra el equipo más goleador, el equipo más goleado y la
    clasificación ordenada de mayor a menor según los puntos.

    Args:
        equipos (dict): Diccionario con las estadísticas de cada equipo.

    Returns:
        None: La función muestra los resultados por pantalla.
    """

  clasificacion = ordenar_clasificacion(equipos)
  mas_goleador = equipo_mas_goleador(equipos)
  mas_goleado = equipo_mas_goleado(equipos)

  print("\n" + "=" * 80)
  print("=" * 28 + "ESTADÍSTICAS DE LA LIGA" + "=" * 29)
  print("=" * 80)

  print(
      f"\nEquipo más goleador: "
      f"{mas_goleador[0]} "
      f"({mas_goleador[1]['goles_favor']} goles)"
  )

  print(
      f"Equipo más goleado: "
      f"{mas_goleado[0]} "
      f"({mas_goleado[1]['goles_contra']} goles)"
  )

  print("\n" + "=" * 105)
  print("=" * 40 + " CLASIFICACIÓN FINAL " + "=" * 44)
  print("=" * 105)

  print(
      f"{'Pos.':<6}"
      f"{'Equipo':<20}"
      f"{'Goles a favor':<16}"
      f"{'Goles en contra':<18}"
      f"{'Ganados':<10}"
      f"{'Empatados':<12}"
      f"{'Perdidos':<10}"
      f"{'Puntos':<8}"
  )

  print("-" * 105)

  posicion = 1

  for nombre, datos in clasificacion:
      print(
          f"{posicion:<6}"
          f"{nombre:<20}"
          f"{datos['goles_favor']:<16}"
          f"{datos['goles_contra']:<18}"
          f"{datos['ganados']:<10}"
          f"{datos['empatados']:<12}"
          f"{datos['perdidos']:<10}"
          f"{datos['puntos']:<8}"
      )
      posicion += 1

  print("=" * 105)
```

```python
def main():
  """
    Ejecuta el programa principal.

    Lee los partidos del archivo, calcula las estadísticas de los
    equipos y muestra los resultados de la liga.

    Returns:
        None: No devuelve ningún valor.
    """
  partidos = leer_partits("/content/drive/MyDrive/Colab Notebooks/phyton_especialidad/historic partits.txt")
  equipos = calcular_estadisticas(partidos)
  mostrar_resultados(equipos)
```
