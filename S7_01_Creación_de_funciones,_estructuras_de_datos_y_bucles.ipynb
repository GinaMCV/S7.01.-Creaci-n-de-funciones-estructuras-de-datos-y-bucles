# **Nivel 1**
## **Ejercicio 1: Selección de Personal**

Desarrollar una herramienta sencilla en Python que actúe como formulario de inscripción y sistema de clasificación inicial.Este formulario debe permitir introducir, para cada candidato/a:

1. Nombre y Apellidos
2. Teléfono de contacto
3. Años de experiencia dentro de la empresa
4. Conocimiento de Catalán (0-3 puntos)
- 0 = sin conocimiento
- 1 = hasta nivel B1
- 2 = nivel B2
- 3 = C1 o superior
5. Dominio de Excel (0-3 puntos)
- 0 = nivel básico
- 1 = uso de funciones
- 2 = tablas dinámicas
- 3 = Visual Basic y Macros
6. Título académico (0-3 puntos)
- 0 = ESO
- 1 = FP
- 2 = Universidad
- 3 = Máster o superior

Asegurándose de que toda la información introducida es correcta y entra dentro de los valores permitidos. A partir de estos datos, el programa deberá calcular una puntuación total aplicando el peso de la experiencia (años * 1.5) y sumando los puntos de catalán, excel y titulación, ya continuación clasificar automáticamente a cada persona según los umbrales establecidos por la empresa:

- Prioridad Alta → puntuación ≥ 12
- Candidato adecuado → 8 ≤ puntuación < 12
- No prioritario → puntuación < 8

Cada vez que alguien completo en el formulario, debe mostrar su puntaje individual. Cuando ya se hayan inscrito exactamente diez personas, la herramienta debe dejar de aceptar nuevos registros y generar un informe final en pantalla, ordenado primero según la prioridad y, en caso de empate, por la puntuación obtenida. Este informe debe incluir el nombre, el teléfono y la puntuación de cada candidato para que el departamento de RRHH pueda contactarles rápidamente.

```python
#Datos para realizar las pruebas

inputs_test = [
    "Laura Serra", "608112233", 6, 3, 3, 3,
    "Clara Gómez", "655221199", 3, 3, 3, 2,
    "Sergio López", "644112211", 3, 2, 2, 2,
    "Candidato Error", "611000000", 5, 7, 3, 2,
    "Marta Ruiz", "699887766", 2, 3, 2, 1,
    "Marc Vidal", "612443322", 6, 3, 3, 3,
    "Júlia Romero", "677998877", 5, 3, 3, 2,
    "Pol Navarro", "699554433", 4, 3, 2, 2,
    "David Ferrer", "677665544", 2, 1, 1, 1,
    "Anna Paredes", "633221144", 1, 2, 1, 0,
    "Aníbal Paredes", "644272727", 1, 2, 1, 0,
    "Núria Casals", "600112200", 8, 2, 3, 3
]


#Funcion que nos permita a partir de la lista inputs_test, simular la entrada de los datos datos
indice = 0
def input_datos():
    global indice
    valor = inputs_test[indice]
    indice += 1
    return valor

#Funcion para obtener los datos a partir de la funcion input_datos, y guardarlos en una variable que los identifique
def obtener_datos():
    nombre = input_datos()
    telefono = input_datos()
    experiencia = input_datos()
    catalan = input_datos()
    excel = input_datos()
    titulacion = input_datos()
    return nombre, telefono, experiencia, catalan, excel, titulacion


#Funcion que permite realizar comprobaciones para que los datos cuenten con la estructura correcta
#Si algún dato es incorrecto, devolvuelve False y si todos los datos son correctos, devolver True.

def validar_datos(nombre, telefono, experiencia, catalan, excel, titulacion):
    if nombre == "": # el nombre no este vacio
        return False
    if len(telefono) != 9: # el telefono tenga 9 digitos
        return False
    if experiencia <= 0: # la experiencia sea mayor o igual que 0
        return False
    if catalan not in range(4): # el nivel de catalan este entre 0 y 3
        return False
    if excel not in range(4): #el nivel de Excel esté entre 0 y 3
        return False
    if titulacion not in range(4): #el nivel de titulación esté entre 0 y 3
        return False
    return True

#funcion para calcular la puntuacion, se calcula el peso de la experiencia, multiplicando los años de experiencia por 1.5 y a esto le sumamos catalan, excel y titulacion
def calcular_puntuacion(experiencia, catalan, excel, titulacion):
    return experiencia * 1.5 + catalan + excel + titulacion


#Funcion para clasificar los candidatos en base a su puntuacion (calcular_puntuacion)
def clasificar_candidato(puntuacion):
    if puntuacion >= 12:
        return "Prioridad alta"     #  si el calculo de la puntuacion es mayor o igual a 12 se clasifica como prioridad alta
    elif puntuacion >= 8:
        return "Candidato adecuado" #  si el calculo de la puntuacion es mayor o igual a 8 se clasifica como candidato adecuado
    else:
        return "No prioritario"    # si el calculo de la puntuacion es inferior a 8 se clasifica como no prioritario

#Funcion para que se muestre el nombre la puntuacion y la clasificasion de cada candidato
def mostrar_individual(nombre, puntuacion, clasificacion):
    print(f"\nCandidato: {nombre}")
    print(f"Puntuación: {puntuacion}")
    print(f"Clasificación: {clasificacion}")


#funcion para mostrar el informe final con el nombre telefono y puntuacion de cada candidato ordenado de mayor a menor en base a la clasificacion o puntuacion
def mostrar_informe(candidatos):
    prioridad = {
        "Prioridad alta": 0,
        "Candidato adecuado": 1,        # se ordenan los candidatos segun la clasificcacion
        "No prioritario": 2
    }
    candidatos.sort(
        key=lambda candidato: (prioridad[candidato["clasificacion"]],     #Si dos candidatos tienen la misma clasificación, ordenarlos según la puntuación obtenida.
                        -candidato["puntuacion"])
    )
    print("\nINFORME FINAL")
    for candidato in candidatos:
        print(
            f'{candidato["nombre"]} - '
            f'{candidato["telefono"]} - '
            f'{candidato["puntuacion"]}'
        )

def main_seleccion():
    candidatos = []      #se crea una lista vacía para almacenar a los candidatos
    while len(candidatos) < 10:           #Mientras haya menos de 10 candidatos válidos
        nombre, telefono, experiencia, catalan, excel, titulacion = obtener_datos() #se llama la funcion para obtener los datos
        if not validar_datos(
            nombre,
            telefono,
            experiencia,
            catalan,                      #se llama la funcion para validar los datos.
            excel,
            titulacion
        ):
            print(f"\nDatos no válidos para {nombre}")
            continue                                  # en caso tal si los datos son incorrectos, mostrar un mensaje con el nombre del canditado y continuar con el siguiente candidato.
        puntuacion = calcular_puntuacion(
            experiencia,
            catalan,                         # se llama la funcion para calcular la puntuación
            excel,
            titulacion
        )
        clasificacion = clasificar_candidato(
            puntuacion
        )                                           #se llama la funcion para clasificar a los candidatos
        candidatos.append(
            {
                "nombre": nombre,
                "telefono": telefono,
                "puntuacion": puntuacion,               #se guarda la informacion en la lista que creamos anteriormente (candidatos)
                "clasificacion": clasificacion
            }
        )
        mostrar_individual(
            nombre,
            puntuacion,                             #se llama la funcion para orenar los canditados
            clasificacion
        )
    mostrar_informe(candidatos)                    #se llama la funcion para mostrar el informe final de los candidatos


main_seleccion()
```
